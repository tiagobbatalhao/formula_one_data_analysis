## Adding distance

```python
from scipy.spatial import KDTree
data_telemetry
```

```python
data_position = pd.read_parquet('../data/gold/telemetry_pos_Y2026R01.parquet')
data_map = pd.read_parquet('../data/circuit_map_Y2026R01S1_Y2026R01S2_Y2026R01S3_Y2026R01S4.parquet')

```

```python
tree_columns = ['coordinate_x', 'coordinate_y']
tree = KDTree(data_map[tree_columns])
aux = tree.query(data_position[tree_columns])
```

```python
data_map
```

```python
for col in ['total_distance_m', 'distance_m', 'coordinate_x', 'coordinate_y', 'coordinate_z']:
    data_position[col + "_circuit"] = data_map.iloc[aux[1]][col].values
```

```python
data_position.merge(
    data_laps.sort_values(by=['time_lap']).iloc[:1][join_on],
    on=join_on, how='inner'
).assign(
    encoding=lambda df: (1.0 * df['distance_m_circuit'] / df['total_distance_m_circuit'])
).assign(
    _cos=lambda df: np.cos(2 * np.pi * df['encoding']),
    _sin=lambda df: np.sin(2 * np.pi * df['encoding']),
)[['timing_from_lap', 'distance_m_circuit', 'encoding', '_cos', '_sin']].sort_values(by=['timing_from_lap'])
```

```python
lap_telemetry = data_telemetry.merge(
    data_laps.sort_values(by=['time_lap']).iloc[:1][join_on],
    on=join_on, how='inner'
)
lap_position = data_position.merge(
    data_laps.sort_values(by=['time_lap']).iloc[:1][join_on],
    on=join_on, how='inner'
)
```

```python
lap_telemetry
```

```python
lap_position
```

```python
_distance = np.interp(
    lap_telemetry['timing_from_lap'],
    lap_position['timing_from_lap'],
    lap_position['distance_m_circuit']
)
total = lap_position['total_distance_m_circuit'].iloc[0]
lap_telemetry['enc_cos'] = np.cos(2 * np.pi * _distance / total)
lap_telemetry['enc_sin'] = np.sin(2 * np.pi * _distance / total)
lap_position['enc_cos'] = np.cos(2 * np.pi * lap_position['distance_m_circuit'] / total)
lap_position['enc_sin'] = np.sin(2 * np.pi * lap_position['distance_m_circuit'] / total)
```
