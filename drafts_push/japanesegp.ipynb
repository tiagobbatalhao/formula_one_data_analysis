```python
import fastf1
import pandas as pd
```

```python
session = fastf1.get_session(2026, 3, 1)
session.load(weather=True, laps=True, messages=True)
```

```python
laps = list(session.laps.iterlaps())
```

```python
car_data = pd.concat([
    x[1].get_car_data() for x in laps
])
```

```python
car_data['DRS'].value_counts()
```

```python
car_data['Speed'].plot.hist(bins=20)
```
