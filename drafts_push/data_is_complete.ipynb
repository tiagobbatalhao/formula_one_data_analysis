```python
import pathlib
import pandas as pd
```

```python
files = list(pathlib.Path("data/bronze").resolve().rglob("*"))
```

```python
def file_properties(fl):
    parts = fl.name.split(".")[-2].split("_")
    df = pd.read_parquet(fl)
    return {
        "session_id": parts[-1],
        "type": "_".join(parts[:-1]),
        "name": fl.name,
        "shape": tuple(df.shape),
    }
```

```python
file_types = pd.DataFrame([file_properties(x) for x in files])
```

```python
aux = file_types[file_types["session_id"].apply(len)==10]
df = (
    aux[["session_id"]].drop_duplicates().assign(k=1)
    .merge(aux[["type"]].drop_duplicates().assign(k=1), on=["k"])
    .merge(aux.assign(exists=lambda df: df['shape'].apply(lambda x: x[0]>0)), on=["session_id", "type"], how="left")
    .fillna({"exists": False})
)
# df = df[(df["session_id"].apply(len)==10) & (df["type"] != "schedule")]
```
