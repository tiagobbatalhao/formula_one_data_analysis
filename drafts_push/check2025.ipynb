```python
data_telemetry = pd.read_parquet('../data/gold/telemetry_car_Y2025R01.parquet')
aux = data_laps.sort_values(by=['time_lap']).iloc[:1]
join_on = ['session_id', 'driver_number', 'lap_number']
```
