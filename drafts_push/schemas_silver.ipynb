```python
import pandas as pd
import numpy as np
from pydantic import BaseModel, field_validator
import pathlib
import collections
from typing import Literal, Optional, List
from datetime import datetime, timedelta, date
```

```python
files = list(pathlib.Path('../data/silver').resolve().rglob('*.parquet'))
def get_type(name):
    return '_'.join(name.split('/')[-1].split('_')[:-1])
```

```python
collections.Counter([get_type(x.name) for x in files])
```

```python
SCHEMA_NAME = "circuit"
```

```python
df = pd.concat([pd.read_parquet(x) for x in files if get_type(x.name)==SCHEMA_NAME][:5])
```
